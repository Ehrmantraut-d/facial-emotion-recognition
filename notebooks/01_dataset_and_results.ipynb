# FER2013 dataset and experiment results
This notebook reads only outputs created by the local pipeline. It does not download data or invent metrics. Quick-mode runs are development-only.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'src' / 'config.py').exists(): ROOT = ROOT.parent
RESULTS = ROOT / 'results'
print('Project:', ROOT)
```

```python
meta = ROOT / 'data' / 'metadata' / 'data_validation.json'
if meta.exists():
    report = json.loads(meta.read_text(encoding='utf-8'))
    display(pd.DataFrame(report['label_counts']).fillna(0).astype(int))
    print('Invalid rows:', report['invalid_count'])
    print('Prepared counts:', report.get('prepared_counts'))
else:
    print('No validated dataset output yet. Configure Kaggle or place the authorized FER2013 CSV, then run the pipeline.')
```

```python
summary = RESULTS / 'experiments_summary.csv'
if summary.exists():
    results = pd.read_csv(summary)
    display(results.sort_values(['mode','macro_f1'], ascending=[True,False]))
else:
    print('No experiment runs saved yet.')
```

## Current model result: FER2013 and mirror averaging

The active inference strategy averages the MLP probabilities from the face crop and its horizontal mirror. Detailed dataset and FER+ consensus results are in `02_ml_results_and_ferplus.ipynb`.

```python
from pathlib import Path
import json
import pandas as pd
full_run = ROOT / 'results' / 'runs' / '20261121T222130Z_full_6297687c'
flip_run = ROOT / 'results' / 'runs' / '20261121T225141Z_flip_average_64697dee'
base_val = pd.read_csv(full_run / 'validation_comparison.csv').query("name == 'mlp_128_64'").iloc[0]
flip_val = json.loads((flip_run / 'validation_flip_average.json').read_text(encoding='utf-8'))
base_test = json.loads((full_run / 'final_test_metrics.json').read_text(encoding='utf-8'))
flip_test = json.loads((flip_run / 'final_test_metrics.json').read_text(encoding='utf-8'))
display(pd.DataFrame([
    {'strategy':'Single view', 'split':'PublicTest', 'accuracy':base_val.accuracy, 'macro_f1':base_val.macro_f1},
    {'strategy':'Mirror probability average', 'split':'PublicTest', 'accuracy':flip_val['accuracy'], 'macro_f1':flip_val['macro_f1']},
    {'strategy':'Single view', 'split':'PrivateTest', 'accuracy':base_test['accuracy'], 'macro_f1':base_test['macro_f1']},
    {'strategy':'Mirror probability average', 'split':'PrivateTest', 'accuracy':flip_test['accuracy'], 'macro_f1':flip_test['macro_f1']},
]).style.format({'accuracy':'{:.2%}','macro_f1':'{:.2%}'}))
```
